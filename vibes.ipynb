```python
# Creating my first machine learing model for predicting the price estimate of a house
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
#Creating a reproducible synthetic dataset
np.random.seed(42)
number_of_houses = 200

square_feet = np.random.randint(
    700, 3501, number_of_houses
).astype(float)
bedrooms = np.random.randint(
    1, 6, number_of_houses
).astype(float)
age = np.random.randint(
    0, 61, number_of_houses
).astype(float)
noise = np.random.normal(
    0, 25000, number_of_houses
).astype(float)

```

```python
#Create synthetic house prices

price = (
    45000000
    + 2000000 * square_feet
    + 1200000 * bedrooms
    - 1000000 * age
    + noise
)
```

```python
# Organize the data inot a table
house = pd.DataFrame({
    "square_feet": square_feet,
    "bedrooms": bedrooms,
    "age": age,
    "price": price.round(2)
})
```

```python
# Separate the features from the target
X = house[["square_feet", "bedrooms", "age"]]
Y = house["price"]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size= 0.20,
    random_state= 42
)
```

```python
# Create and train the model
model = LinearRegression()
model.fit(X_train, Y_train)
predictions = model.predict(X_test)
print(predictions)
```

```python
# Compare several predictions with the actual price
results = pd.DataFrame({
    "actual_price": Y_test.values,
    "predicted_price": predictions.round(2)
})

print("\nACTUAL VS. PREDICTED PRICES")
print(results.head(10))
```

```python
#Predict the price of a new house
new_house = pd.DataFrame({
    "square_feet": [1800],
    "bedrooms": [3],
    "age": [10]
})

predicted_price = model.predict(new_house)
print(predicted_price)
```
