```python
#Broadcasting
df["Salary"] = df["Salary"] + 5000

df["Salary"]
```

```python
df["Promoted Salary"] = df["Salary"]* 10
df
```

```python
# Data Cleaning

df.isnull().sum()
```

```python
# Used to drop Null/NaN values 

df.dropna(how= "all") # Removes row if only all the values are Null/Naa
df.dropna(how="any") # Removes values that have any Null values in a row 
```

```python
df["Age"].fillna(df["Age"].mean())
```

```python
df["Salary"].fillna(df["Salary"].median())
```

```python
df["Age"].fillna(method = "bfill")
```

```python
df["Name"].replace("Charlie", "Rose")
```

```python
#  Duplicate values

df = df.drop_duplicates()
```

```python
# Invalid values
df["Promoted Salary"] = df["Promoted Salary"].apply(lambda x:x/10 if x > 650000 else x )
df
```
